```python
import pandas as pd
```

```python
df = pd.read_csv("pakistani_ecommerce_dataset.csv")
df
```

Q1: Which is the most popular product category?

```python
df["product_category"].value_counts()
```

Ans: The most popular category from the above analysis is Home Decore.

Q2: What is the average order value?

```python
print(f"Average Order Value: {df['order_value'].mean():.2f}")
```

Q3: Which city has the highest number of orders?

```python
df["city"].value_counts()
```

Ans: Karachi has the highest number of orders

Q4: What is the most used payment method?

```python
df["payment_method"].value_counts()
```

Ans: The most used payment method is Cash on Delivery

Q5: Draw a bar chart showing average order value by city. 

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
city_avg = df.groupby("city")["order_value"].mean().round(2)
city_avg
```

```python
plt.figure(figsize=(12, 8))
sns.barplot(x=city_avg.index, hue=city_avg, y=city_avg.values, palette="viridis")
plt.title("Average order values of cities")
plt.xlabel("City")
plt.ylabel("Order Value")
plt.xticks(rotation=45)
plt.show()
```

Q1: What is the average number of study hours?

Q3: Is there a relationship between screen time and sleep hours? 

```python
print(f"The correlaton between screen time and sleep are is: {df2["screen_time"].corr(df2["sleep_hours"]).round(4)}")
```
