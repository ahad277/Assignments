```python
import pandas as pd
```

```python
df = pd.read_csv("student_tea_snacks _1_.csv")
df
```

Q1: What is the most preferred tea type?

```python
df["tea_type"].value_counts()
```

Ans: Green tea is the most preffered tea among all.

Q2: What is the average number of cups per gender?

```python
df.groupby("gender")["cups_per_day"].mean().round(2)
```

Q3: Make a bar chart of snack choices

```python
import matplotlib.pyplot as plt
```

```python
snack_counts = df["snack_choice"].value_counts()
plt.figure(figsize=(10, 7))
snack_counts.plot(kind="bar", title="Snack Choices", xlabel="Snack", ylabel="Count")
plt.tight_layout()
plt.savefig("snack_bar_chart.png")
plt.xticks(rotation=45)
plt.show()
```

Q4: Compare tea preference by age group

```python
df.groupby("prefers_tea")["age_group"].value_counts().to_frame()
```
